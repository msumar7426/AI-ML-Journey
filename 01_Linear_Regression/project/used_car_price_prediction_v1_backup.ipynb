```python
car_specs=['manufacturer', 'variant', 'year', 'mileage', 'fuelType', 'transmission', 'engine']
near_duplicates=df[df.duplicated(subset=car_specs, keep=False)].sort_values(by=
    ['manufacturer', 'variant', 'year', 'mileage'])[car_specs+['listingPrice','name']].head(20)

near_duplicates
```

```python
exact_dupes = df.duplicated(keep=False).sum()
near_dupes = df.duplicated(subset=car_specs, keep=False).sum()

print(f"Exact duplicates (all columns): {exact_dupes}")
print(f"Near duplicates (specs only): {near_dupes}")
```

```python
# Step 1: Ask Question 1 - "Who has identical car specs?"
near_dup_mask = df.duplicated(subset=car_specs, keep=False)

# Step 2: Ask Question 2 - "Who is a perfect clone?"
exact_dup_mask = df.duplicated(keep=False)

# Step 3: Combine  "Spec-twins that are NOT perfect clones"
final_mask = near_dup_mask & ~exact_dup_mask

# Step 4: Use the mask to filter the dataframe
df[final_mask].sort_values(
    by=['manufacturer', 'variant', 'year', 'mileage']
)[car_specs + ['listingPrice', 'name']].head(20)
```

```python
# Before cleaning
print(f"Original shape: {df.shape}")

# Drop duplicates based on car specs only (handles both exact AND near duplicates)
""" keep='first' does: For every group of rows with identical specs,
keep the one with the lowest original index and throw away the rest."""
df_clean = df.drop_duplicates(subset=car_specs, keep='first')

# After cleaning
print(f"Clean shape: {df_clean.shape}")
print(f"Rows dropped: {df.shape[0] - df_clean.shape[0]}")
```

```python
# Same approach
print("First 20 unique mileages:")
print(df_clean['mileage'].unique()[:20])

print("\nMileage value counts tail:")
print(df_clean['mileage'].value_counts().tail(20))
```
